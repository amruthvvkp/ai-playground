This notebook creates a synthetic data for testing customer feedback analysis.


```python
import pandas as pd
from pathlib import Path
from faker import Faker
import random

# Initialize Faker
fake = Faker()

# Function to generate synthetic feedback text
def generate_feedback():
    feedback_templates = [
        "The application crashes whenever I try to {}.",
        "Great user interface, but the {} is slow.",
        "I found a bug in the {}; it keeps {}.",
        "Customer support is responsive, but the {} is confusing.",
        "The system flagged a false positive on my {}.",
        "I am unable to {} through the app.",
        "Overall, a good experience, but I wish the {} were easier to find.",
        "The tool's {} feature is very accurate, but it takes too long to {}.",
        "I experienced a {}, and the support team didn't help much.",
        "Great tool for {}, but it lacks integration with third-party apps.",
    ]
    actions = ["upload a file", "login", "reporting tool", "security dashboard", "security scan", "reset my password", "security settings", "threat detection", "data breach", "cybersecurity"]
    issues = ["upload a file", "login", "reporting tool", "security dashboard", "security scan", "reset my password", "security settings", "analyze", "data breach", "cybersecurity"]
    return random.choice(feedback_templates).format(random.choice(actions), random.choice(issues))

# List of priorities
priorities = ["High", "Medium", "Low"]

# Generate synthetic data
data = {
    "text": [generate_feedback() for _ in range(300)],
    "priority": [random.choice(priorities) for _ in range(300)],
}

# Create a DataFrame
df = pd.DataFrame(data)

# Get current path of the notebook
notebook_path = Path.cwd()

# Save synthetic data to CSV
train_path = notebook_path / "synthetic_train_data.csv"
test_path = notebook_path / "synthetic_test_data.csv"

# Save the data to CSV for both train and test purposes
df.to_csv(train_path, index=False)
df.to_csv(test_path, index=False)

train_path, test_path
```
